```python
from langchain.agents import create_agent
from langchain_core.tools import tool
from pydantic import BaseModel, Field

@tool
def get_weather(city: str) -> str:
    """根据输入城市查询天气
    Args:
        city: 要查询天气的城市名称，例如： 北京
    """
    return "挺好"

class SupportReply(BaseModel):
    """订单客服最终返回的结构化应答"""
    content: str = Field(description="给用户的自然语言回复")
    order_status: str | None = Field(None, description="订单状态：运输中/已签收/异常")
    need_transfer_human: bool = Field(description="是否需要转人工")

agent=create_agent(
    mode="openai:xxx",
    tools=[get_weather],
    system_prompt=(
        "你是订单客服，xxxx"
    ),
    response_format=SupportReply 
)

result=agent.invoke({
    "messages":[{"role":"user","content":"订单 A100 到哪了？"}]
})
reply: SupportReply= result["structured_response"]
```
